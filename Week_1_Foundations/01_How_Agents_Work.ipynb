# How Agents Work

*Notebook 01*

A basic agent combines a model with instructions.

`Agent` sets the name, instructions, and model.

`Runner.run()` executes the agent.

`result.final_output` holds the response.


---

## 🔧 Setup

```python
from pathlib import Path
from dotenv import load_dotenv
load_dotenv(dotenv_path=Path("..") / ".env")

from agents import Agent, Runner

MODEL = "gpt-5-mini"       # Course baseline: fast, lower-cost tasks
REASONING_MODEL = "gpt-5"  # Planning and orchestration (Lesson 29)

print("✅ Ready!")
```

---

## 🚀 How an Agent Works

```python
question = "What is Python?"

agent = Agent(
    name="Assistant",
    instructions="You are a helpful assistant. "
                 "Answer briefly without headings or bullet lists.",
    model=MODEL
)

result = await Runner.run(agent, input=question)

print(result.final_output)
```

### 💡 Key Takeaway

A basic agent run separates configuration, execution, and final output.

---

## 🧪 Part 1: Instructions Shape Behavior

Both agents use the same model and answer the same `question`.

Only their instructions change.


#### Run Concise Agent

```python
concise_agent = Agent(
    name="Concise Assistant",
    instructions="Answer every question in exactly one sentence.",
    model=MODEL
)

concise_result = await Runner.run(concise_agent, input=question)

print("📌 Concise:")
print(concise_result.final_output)
```

#### Run Friendly Agent

```python
friendly_agent = Agent(
    name="Friendly Assistant",
    instructions="Explain concepts to a 10-year-old using a fun analogy. "
                 "Keep the answer to one short paragraph.",
    model=MODEL
)

friendly_result = await Runner.run(friendly_agent, input=question)

print("📌 Friendly:")
print(friendly_result.final_output)
```

### 💡 Key Takeaway

Same model and question. Different instructions produce different answers.

## 🎯 Key Takeaways

**Agent runs separate configuration, execution, and output:**

- `Agent`: configures the name, instructions, and model.

- `Runner.run()`: executes the agent and returns a result.

- `result.final_output`: holds the final response.
<br>
<br>

**Instructions change behavior without changing the model:**

- Use `instructions` to shape tone, audience, and response length.
<br>
<br>

**Tracing gives you a starting point for debugging:**

- Runs are recorded by default.

- Open the Traces dashboard when behavior surprises you.
