# How Agents Work

*Notebook 01*

At its simplest, an agent combines a model with instructions for how to respond.

This lesson uses three SDK pieces:

- `Agent`: configures the name, instructions, and model.

- `Runner.run()`: executes the agent and returns a result.

- `result.final_output`: holds the response text.


---

## 🔧 Setup

```python
from pathlib import Path
from dotenv import load_dotenv
load_dotenv(dotenv_path=Path("..") / ".env")

from agents import Agent, Runner

MODEL = "gpt-5-mini"       # Course baseline: fast, lower-cost tasks
REASONING_MODEL = "gpt-5"  # Planning and orchestration (Lesson 29)

print("✅ Ready!")
```

---

## 🚀 How an Agent Works

Run the cell below to see those three pieces in action.

```python
message = "What is Python?"

agent = Agent(
    name="Assistant",
    instructions="You are a helpful assistant. "
                 "Answer briefly without headings or bullet lists.",
    model=MODEL
)

result = await Runner.run(agent, input=message)

print(result.final_output)
```

### 💡 Key Takeaway

A basic agent run separates configuration, execution, and final output.

---

## 🧪 Part 1: Instructions Shape Behavior

Both agents answer the same `message`.

Only their instructions change.

#### Run Concise Agent

```python
agent_concise = Agent(
    name="Concise Assistant",
    instructions="Answer every question in exactly one sentence.",
    model=MODEL
)

concise_result = await Runner.run(agent_concise, input=message)

print("📌 Concise:")
print(concise_result.final_output)
```

#### Run Friendly Agent

```python
friendly_instructions = (
    "Explain concepts to a 10-year-old using a fun analogy. "
    "Keep the answer to one short paragraph."
)

agent_friendly = Agent(
    name="Friendly Assistant",
    instructions=friendly_instructions,
    model=MODEL
)

friendly_result = await Runner.run(agent_friendly, input=message)

print("📌 Friendly:")
print(friendly_result.final_output)
```

### 💡 Key Takeaway

Same model and question. Different instructions produce different answers.
